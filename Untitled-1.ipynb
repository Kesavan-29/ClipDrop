```python
from pathlib import Path
import subprocess

from pytubefix import YouTube
import imageio_ffmpeg

url = "https://youtu.be/JtaOmwnR6AM"
yt = YouTube(url)

video = (
    yt.streams
    .filter(only_video=True, file_extension="mp4")
    .order_by("resolution")
    .desc()
    .first()
)
audio = (
    yt.streams
    .filter(only_audio=True)
    .order_by("abr")
    .desc()
    .first()
)

if video is None or audio is None:
    raise RuntimeError("This video does not provide both downloadable video and audio streams.")

output_dir = Path.cwd() / "youtube_download"
output_dir.mkdir(exist_ok=True)
video_path = Path(video.download(output_path=str(output_dir), filename_prefix="video_"))
audio_path = Path(audio.download(output_path=str(output_dir), filename_prefix="audio_"))
output_path = output_dir / "download.mp4"

subprocess.run(
    [
        imageio_ffmpeg.get_ffmpeg_exe(),
        "-y",
        "-i", str(video_path),
        "-i", str(audio_path),
        "-map", "0:v:0",
        "-map", "1:a:0",
        "-c:v", "copy",
        "-c:a", "aac",
        str(output_path),
    ],
    check=True,
)

print(f"Download complete with audio: {output_path}")
```
